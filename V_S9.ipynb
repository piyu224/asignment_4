```python
# 1.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "likes": [100, 250, 400, 550, 700, 850, 1000, 1200, 1400, 1600],
    "comments": [10, 20, 35, 40, 55, 65, 80, 95, 110, 125]
}

df = pd.DataFrame(data)

sns.regplot(data=df, x="likes", y="comments")

plt.xlabel("Number of Likes")
plt.ylabel("Number of Comments")
plt.title("Instagram Likes vs Comments")
plt.show()
```

```python
# 2.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "price": [120, 150, 200, 250, 180, 300, 350, 220, 400, 275,
              160, 450, 320, 240, 380],
    
    "delivery_time": [20, 25, 30, 35, 28, 40, 45, 32, 50, 38,
                      24, 55, 42, 34, 48]
}

df = pd.DataFrame(data)

sns.lmplot(data=df, x="price", y="delivery_time")

plt.xlabel("Food Order Price")
plt.ylabel("Delivery Time (minutes)")
plt.title("Food Order Price vs Delivery Time")
plt.show()
```

```python
# 3.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "songs": [5, 10, 15, 20, 25, 30, 35, 40, 45, 50],
    "duration": [18, 35, 52, 70, 85, 105, 118, 145, 160, 180]
}

df = pd.DataFrame(data)

sns.regplot(
    data=df,
    x="songs",
    y="duration",
    lowess=True
)

plt.xlabel("Number of Songs")
plt.ylabel("Playlist Duration (minutes)")
plt.title("Spotify Songs vs Playlist Duration")
plt.show()
```

```python
# 5.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "price": [5000, 8000, 12000, 15000, 18000,
              22000, 25000, 30000, 35000, 40000],
    
    "reviews": [500, 800, 1200, 1500, 1800,
                2200, 2500, 3000, 3500, 4000]
}

df = pd.DataFrame(data)

sns.lmplot(
    data=df,
    x="price",
    y="reviews",
    scatter_kws={"color": "red"}
)

plt.xlabel("Product Price (₹)")
plt.ylabel("Number of Reviews")
plt.title("Flipkart Product Price vs Number of Reviews")

plt.show()
```
