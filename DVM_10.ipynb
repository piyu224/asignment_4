```python
# 1.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

users = [1200, 1350, 1280, 1500, 1600, 1750,
         1680, 1800, 1950, 2100, 2050, 2300]

plt.plot(months, users, marker='o')

highest = max(users)
highest_month = months[users.index(highest)]

plt.annotate(
    "Highest Users",
    xy=(highest_month, highest),
    xytext=(highest_month, highest + 200),
    arrowprops=dict(arrowstyle="->")
)

plt.title("Monthly Active Users - Music Streaming App")
plt.xlabel("Month")
plt.ylabel("Daily Active Users")
plt.grid(True)
plt.show()
```

```python
# 2.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

orders = [850, 920, 780, 1100, 1050, 980,
          1150, 1080, 760, 1200, 1250, 1180]

plt.bar(months, orders)

lowest = min(orders)
lowest_month = months[orders.index(lowest)]

plt.annotate(
    "Lowest Orders",
    xy=(lowest_month, lowest),
    xytext=(lowest_month, lowest + 250),
    arrowprops=dict(arrowstyle="->")
)

plt.title("Monthly Orders - Food Delivery App")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.show()
```

```python
# 3.

import matplotlib.pyplot as plt

payment_methods = ["UPI", "Card", "Wallet", "Cash"]
payments = [45, 30, 15, 10]

plt.pie(
    payments,
    labels=payment_methods,
    autopct="%1.1f%%"
)

plt.title("Payment Methods - E-commerce App")

plt.legend(
    payment_methods,
    loc="lower right"
)

plt.show()
```

```python
# 4.

import matplotlib.pyplot as plt

movies = ["Movie A", "Movie B", "Movie C", "Movie D", "Movie E"]
ratings = [7.5, 8.2, 9.1, 7.8, 8.5]
reviews = [1200, 1800, 2500, 1500, 2200]

plt.scatter(reviews, ratings)

highest = max(ratings)
highest_index = ratings.index(highest)

plt.annotate(
    movies[highest_index] + " - Highest Rating",
    xy=(reviews[highest_index], highest),
    xytext=(reviews[highest_index] - 800, highest - 0.5),
    bbox=dict(boxstyle="round", facecolor="white", edgecolor="black")
)

plt.title("Movie Ratings vs Number of Reviews")
plt.xlabel("Number of Reviews")
plt.ylabel("Rating")
plt.grid(True)
plt.show()
```
