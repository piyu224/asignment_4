```python
# 1.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [4500, 6000, 5200, 7000, 6500, 8000, 7500]

plt.plot(days, steps, color="green", linewidth=2)

plt.xlabel("Days")
plt.ylabel("Number of Steps")
plt.title("Daily Steps")

plt.show()
```

```python
# 2.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
users = [100, 120, 150, 180, 200, 230]

plt.plot(
    months,
    users,
    marker="o",
    markerfacecolor="red"
)

plt.xlabel("Month")
plt.ylabel("Active Users")
plt.title("Monthly Active Users")

plt.show()
```

```python
# 3.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

zomato_orders = [10, 15, 12, 18, 20]
swiggy_orders = [8, 12, 16, 14, 22]

plt.plot(
    months,
    zomato_orders,
    color="red",
    linestyle="-",
    marker="o",
    label="Zomato"
)

plt.plot(
    months,
    swiggy_orders,
    color="orange",
    linestyle="--",
    marker="o",
    label="Swiggy"
)

plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.title("Zomato vs Swiggy Orders")

plt.legend()
plt.grid(True)

plt.show()
```

```python
# 4.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
data_usage = [8, 10, 12, 15, 13, 18]

plt.plot(
    months,
    data_usage,
    marker="o",
    color="blue",
    linewidth=2
)

plt.title("Monthly Data Usage Trend")
plt.xlabel("Month")
plt.ylabel("Data Used (GB)")

plt.grid(True)

plt.show()
```
