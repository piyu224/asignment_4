```python
# 1.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
users = [1200, 1350, 1280, 1500, 1600, 1800, 1700]

plt.figure(figsize=(10, 4))

plt.plot(days, users, marker="o")

plt.xlabel("Days")
plt.ylabel("Daily Active Users")
plt.title("Music Streaming App - Daily Active Users")

plt.show()
```

```python
# 2.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

zomato = [8, 10, 7, 12, 15, 18, 14]
swiggy = [6, 9, 8, 10, 13, 16, 12]

plt.plot(days, zomato, linestyle="--", marker="o", label="Zomato")
plt.plot(days, swiggy, linestyle=":", marker="o", label="Swiggy")

plt.xlabel("Days")
plt.ylabel("Number of Orders")
plt.title("Zomato vs Swiggy Orders")

plt.legend()

plt.show()
```

```python
# 3.

import matplotlib.pyplot as plt

days = ["Day 1", "Day 2", "Day 3", "Day 4", "Day 5",
        "Day 6", "Day 7", "Day 8", "Day 9", "Day 10"]

sales = [120, 150, 135, 180, 200, 175, 220, 240, 210, 260]

plt.style.use("ggplot")

plt.plot(days, sales, color="orange", marker="o")

plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Flipkart Daily Sales")

plt.show()
```

```python
# 4.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
posts = [5, 8, 6, 10, 12, 15, 9]

colors = ["#833AB4","#C13584","#E1306C","#FD1D1D","#F56040","#F77737","#FCB045"]

plt.bar(days, posts, color=colors)

plt.xlabel("Days")
plt.ylabel("Number of Posts")
plt.title("Instagram Daily Posts")

plt.show()
```

```python
# 5.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
posts = [5, 8, 6, 10, 12, 15, 9]

plt.style.use("dark_background")

plt.bar(days, posts, color="cyan")

plt.xlabel("Days")
plt.ylabel("Number of Posts")
plt.title("Instagram Daily Posts - Dark Theme")

plt.show()
```
