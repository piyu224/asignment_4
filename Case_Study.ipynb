```python
# 1.

import pandas as pd

bank_df = pd.read_csv("Documents/data.csv")

print(bank_df.head(10))

print(bank_df.info())
```

```python
# 2.

import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(data=bank_df, x="y")

plt.xlabel("Term Deposit Subscription")
plt.ylabel("Number of Customers")
plt.title("Customers Who Subscribed to Term Deposit")

plt.show()
```

```python
# 3.

sns.boxplot(data=bank_df, x="y", y="age")

plt.xlabel("Term Deposit Subscription")
plt.ylabel("Age")
plt.title("Age vs Term Deposit Subscription")

plt.show()
```

```python
# 4.

education_percentage = (
    bank_df.groupby("education")["y"]
    .apply(lambda x: (x == "yes").mean() * 100)
    .reset_index(name="Subscription_Percentage")
)

print(education_percentage)
```

```python
# 5.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

bank_df = pd.read_csv("Documents/data.csv")

bank_df.columns = bank_df.columns.str.strip().str.lower()

job_percentage = (
    bank_df.groupby("job")["y"]
    .apply(lambda x: (x == "yes").mean() * 100)
    .reset_index(name="Subscription_Percentage")
)

sns.barplot(
    data=job_percentage,
    x="job",
    y="Subscription_Percentage"
)

plt.xlabel("Job")
plt.ylabel("Subscription Percentage (%)")
plt.title("Term Deposit Subscription by Job")

plt.xticks(rotation=45)
plt.show()
```
