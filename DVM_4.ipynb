```python
# 1.

import matplotlib.pyplot as plt

apps = ["Instagram", "YouTube", "WhatsApp", "Zomato", "Spotify"]
time = [25, 30, 20, 10, 15]

plt.pie(
    time,
    labels=apps,
    autopct="%1.1f%%"
)

plt.title("Daily App Usage")
plt.show()
```

```python
# 2.

import matplotlib.pyplot as plt

categories = ["Food Delivery", "Shopping", "Entertainment", "UPI Payments"]
spending = [3000, 5000, 2000, 4000]

explode = [0, 0.1, 0, 0]

plt.pie(
    spending,
    labels=categories,
    autopct="%1.1f%%",
    explode=explode
)

plt.title("Monthly Online Spending")
plt.show()
```

```python
# 3.

import matplotlib.pyplot as plt

categories = ["Food Delivery", "Shopping", "Entertainment", "UPI Payments"]
spending = [3000, 5000, 2000, 4000]

explode = [0, 0.1, 0, 0]

fig, ax = plt.subplots()

ax.pie(
    spending,
    labels=categories,
    autopct="%1.1f%%",
    explode=explode
)

circle = plt.Circle((0, 0), 0.5, color="white")

ax.add_artist(circle)

ax.set_title("Monthly Online Spending")
plt.show()
```

```python
# 4.

import matplotlib.pyplot as plt

teams = ["CSK", "MI", "RCB", "KKR"]
followers = [15, 14, 13, 8]

fig, ax = plt.subplots()

ax.pie(
    followers,
    labels=teams,
    autopct="%1.1f%%"
)

circle = plt.Circle((0, 0), 0.5, color="white")
ax.add_artist(circle)

ax.set_title("IPL Teams Instagram Followers Share")

plt.show()
```
