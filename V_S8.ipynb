```python
# 1.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "user": ["User1", "User2", "User3", "User4", "User5",
             "User6", "User7", "User8", "User9", "User10",
             "User11", "User12", "User13", "User14", "User15",
             "User16", "User17", "User18", "User19", "User20"],
    
    "followers": [500, 1200, 800, 2500, 1500, 3000, 4500, 2200, 6000, 3500,
                  7000, 900, 1800, 5000, 2800, 6500, 4000, 7500, 3200, 5500],
    
    "posts": [20, 45, 30, 80, 55, 100, 120, 70, 150, 90,
              180, 35, 60, 130, 75, 160, 110, 200, 85, 140]
}

df = pd.DataFrame(data)

sns.scatterplot(data=df, x="followers", y="posts")

plt.xlabel("Number of Followers")
plt.ylabel("Number of Posts")
plt.title("Instagram Followers vs Number of Posts")
plt.show()
```

```python
# 2.

import seaborn as sns
import matplotlib.pyplot as plt
import random

days = list(range(1, 31))

orders = [random.randint(50, 200) for i in range(30)]

sns.lineplot(x=days, y=orders, marker="o")

plt.xlabel("Day of Month")
plt.ylabel("Number of Food Orders")
plt.title("Daily Zomato Food Orders")
plt.show()
```

```python
# 3.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "movie": ["Movie1", "Movie2", "Movie3", "Movie4", "Movie5",
              "Movie6", "Movie7", "Movie8", "Movie9", "Movie10",
              "Movie11", "Movie12", "Movie13", "Movie14", "Movie15"],
    
    "rating": [8.2, 7.5, 8.8, 6.9, 7.8,
               8.5, 7.2, 9.0, 6.8, 8.1,
               7.7, 8.9, 7.0, 8.4, 7.6],
    
    "reviews": [1200, 800, 1500, 600, 900,
                1300, 700, 1800, 500, 1100,
                850, 1600, 650, 1400, 950],
    
    "genre": ["Action", "Comedy", "Drama", "Action", "Comedy",
              "Drama", "Action", "Drama", "Comedy", "Action",
              "Comedy", "Drama", "Action", "Comedy", "Drama"]
}

df = pd.DataFrame(data)

sns.relplot(
    data=df,
    x="rating",
    y="reviews",
    hue="genre"
)

plt.xlabel("Movie Rating")
plt.ylabel("Number of Reviews")
plt.show()
```

```python
# 4.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "youtube_hours": [1, 2, 3, 4, 5, 2, 6, 3, 7, 4,
                      8, 5, 9, 6, 2, 7, 3, 8, 4, 10,
                      5, 6, 7, 9, 8],
    
    "exam_score": [95, 92, 88, 85, 82, 90, 78, 87, 75, 83,
                   70, 80, 68, 76, 91, 73, 86, 71, 84, 65,
                   81, 77, 74, 69, 72]
}

df = pd.DataFrame(data)

sns.regplot(data=df, x="youtube_hours", y="exam_score")

plt.xlabel("Hours Spent on YouTube")
plt.ylabel("Exam Score")
plt.title("YouTube Usage vs Exam Scores")
plt.show()
```
