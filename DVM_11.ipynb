```python
# 1.

import matplotlib.pyplot as plt

restaurants = ["Restaurant A", "Restaurant B", "Restaurant C", "Restaurant D", "Restaurant E"]
orders = [120, 180, 150, 200, 130]

plt.bar(restaurants, orders)

plt.title("Zomato Restaurant Orders")
plt.xlabel("Restaurants")
plt.ylabel("Number of Orders")

plt.xticks(rotation=20)

plt.savefig("zomato_orders.png", dpi=150)
plt.show()
```

```python
# 2.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [5000, 6500, 7000, 5500, 8000, 9000, 7500]

plt.plot(days, steps, marker="o")

plt.title("Daily Step Count")
plt.xlabel("Day")
plt.ylabel("Steps")

plt.grid(True)

plt.savefig("weekly_steps.svg")
plt.show()
```

```python
# 3.

import matplotlib.pyplot as plt

apps = ["Instagram", "YouTube", "WhatsApp", "Spotify"]
time = [3, 2.5, 1.5, 1]

plt.pie(time, labels=apps, autopct="%1.1f%%")

plt.title("Daily App Usage")

plt.savefig("app_usage.pdf")
plt.show()
```

```python
# 5.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [100, 150, 130, 180, 200, 250]

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid(True)

# Save PDF
plt.savefig("sales_200dpi.pdf", dpi=200)

plt.show()
```
