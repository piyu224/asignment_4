```python
# 1.

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)

data = {
    "delivery_time": np.random.randint(15, 60, 20),
    "order_value": np.random.randint(100, 1000, 20),
    "tip_amount": np.random.randint(10, 100, 20),
    "rating": np.round(np.random.uniform(1, 5, 20), 1),
    "delivery_distance": np.round(
        np.random.uniform(1, 15, 20), 2
    )
}

df = pd.DataFrame(data)

print("Food Delivery Data:")
print(df)

correlation_matrix = df.corr()

print("\nCorrelation Matrix:")
print(correlation_matrix)
```

```python
# 2.

plt.figure(figsize=(8, 6))

sns.heatmap(correlation_matrix,annot=True,cmap="coolwarm",fmt=".2f",linewidths=0.5)

plt.title("Food Delivery Correlation Heatmap")
plt.show()
```

```python
# 3.

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))

sns.heatmap(correlation_matrix,annot=True,cmap="coolwarm",fmt=".2f",linewidths=0.5,center=0)

plt.title("Food Delivery Correlation Heatmap - Coolwarm")
plt.show()

corr = correlation_matrix.to_numpy().copy()

np.fill_diagonal(corr, np.nan)

corr = pd.DataFrame(
    corr,
    index=correlation_matrix.index,
    columns=correlation_matrix.columns
)

negative_pair = corr.stack().idxmin()
negative_value = corr.stack().min()

print("Strongest Negative Correlation:")
print(negative_pair)
print("Correlation Value:", round(negative_value, 2))
```

```python
# 4.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Documents/flipkart_mobiles.csv")

df.columns = df.columns.str.strip().str.lower()

print(df.columns)

data = df[["price", "rating", "ram", "battery"]]

correlation_matrix = data.corr(numeric_only=True)

print(correlation_matrix)

sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm")
plt.title("Flipkart Mobiles Correlation Heatmap")
plt.show()
```
