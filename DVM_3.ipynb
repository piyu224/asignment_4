```python
# 1.

import matplotlib.pyplot as plt

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [120, 150, 80, 60, 100]

colors = ["red", "orange", "blue", "green", "purple"]

plt.bar(apps, orders, color=colors)

plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Orders")
plt.title("Daily Orders by Food Delivery Apps")

plt.show()
```

```python
# 3.

import matplotlib.pyplot as plt

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [120, 150, 80, 60, 100]

colors = ["red", "orange", "blue", "green", "purple"]

bars = plt.bar(apps, orders, color=colors)

plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Orders")
plt.title("Daily Orders by Food Delivery Apps")

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        str(height),
        ha="center",
        va="bottom"
    )

plt.show()
```

```python
# 4.

import matplotlib.pyplot as plt

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [120, 150, 80, 60, 100]

colors = ["red", "orange", "blue", "green", "purple"]

# Thinner bars
plt.bar(apps, orders, color=colors, width=0.5)

plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Orders")
plt.title("Daily Orders by Food Delivery Apps")

plt.show()
```
