```python
# 1.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Documents/flipkart_products.csv")

sns.jointplot(data=df, x="Rating", y="Price", kind="scatter")

plt.show()
```

```python
# 4.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Documents/zomato_reviews.csv")

sns.jointplot(data=df,x="rating",y="reviews",kind="scatter")
plt.show()

sns.jointplot(data=df,x="rating",y="reviews",kind="hex")
plt.show()

sns.jointplot(data=df,x="rating",y="reviews",kind="kde")
plt.show()
```
