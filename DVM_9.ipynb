```python
# 1.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

zomato = [8, 10, 7, 12, 15, 18, 14]
swiggy = [6, 9, 8, 10, 13, 16, 12]

plt.bar(days, zomato, label="Zomato")
plt.bar(days, swiggy, bottom=zomato, label="Swiggy")

plt.xlabel("Days")
plt.ylabel("Number of Orders")
plt.title("Zomato and Swiggy Food Orders")
plt.legend()

plt.show()
```

```python
# 2.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify = [1200, 1350, 1400, 1500, 1600, 1800, 1700]
youtube = [1000, 1200, 1300, 1250, 1450, 1600, 1550]

plt.plot(days, spotify, label="Spotify")
plt.fill_between(days, spotify, alpha=0.5)

plt.plot(days, youtube, label="YouTube")
plt.fill_between(days, youtube, alpha=0.5)

plt.xlabel("Days")
plt.ylabel("Daily Active Users")
plt.title("Spotify and YouTube Daily Active Users")
plt.legend()

plt.show()
```

```python
# 3.

import matplotlib.pyplot as plt

days = ["Day 1", "Day 2", "Day 3", "Day 4", "Day 5",
        "Day 6", "Day 7", "Day 8", "Day 9", "Day 10"]

delivery_time = [35, 38, 32, 40, 36, 42, 39, 34, 37, 41]

std_dev = [3, 4, 2, 5, 3, 4, 3, 2, 4, 3]

plt.errorbar(days, delivery_time, yerr=std_dev, fmt="o-",
             capsize=5)

plt.xlabel("Days")
plt.ylabel("Delivery Time (Minutes)")
plt.title("Flipkart Delivery Time with Standard Deviation")

plt.show()
```

```python
# 4.

import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime, timedelta

dates = [datetime(2026, 6, 1) + timedelta(days=i) for i in range(14)]

balance = [2500, 2300, 2450, 2100, 2250, 2400, 2200,
           2600, 2500, 2700, 2550, 2800, 2900, 2750]

plt.plot(dates, balance, marker="o")

plt.xlabel("Date")
plt.ylabel("Wallet Balance (₹)")
plt.title("Paytm Wallet Balance")

plt.gca().xaxis.set_major_formatter(mdates.DateFormatter("%d %b"))

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# 5.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify = [1200, 1350, 1400, 1500, 1600, 1800, 1700]
youtube = [1000, 1200, 1300, 1250, 1450, 1600, 1550]

fig, ax = plt.subplots()

ax.plot(days, spotify, label="Spotify")
ax.fill_between(days, spotify, alpha=0.5)

ax.plot(days, youtube, label="YouTube")
ax.fill_between(days, youtube, alpha=0.5)

ax.set_title("Music & Video App Insights")
ax.set_xlabel("Day of Week")
ax.set_ylabel("Daily Active Users")

ax.grid(True)

ax.legend()

plt.show()
```
