```python
# 1.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [4500, 6000, 5200, 7000, 6500, 8000, 7500]

plt.plot(days, steps, marker="o")

plt.xlabel("Days")
plt.ylabel("Number of Steps")
plt.title("Daily Steps for a Week")
plt.grid(True)

plt.show()
```

```python
# 2.

import matplotlib.pyplot as plt

posts = ["Post 1", "Post 2", "Post 3", "Post 4", "Post 5",
         "Post 6", "Post 7", "Post 8", "Post 9", "Post 10"]

likes = [120, 180, 150, 220, 300, 250, 350, 280, 400, 320]

plt.plot(posts, likes, marker="o")

plt.xlabel("Instagram Posts")
plt.ylabel("Number of Likes")
plt.title("Instagram Likes on Last 10 Posts")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
# 3.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify_hours = [2, 3, 2.5, 4, 3, 5, 4]
youtube_hours = [1.5, 2, 2.5, 1, 3, 4, 3.5]

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot(days, spotify_hours, marker="o")
axes[0].set_xlabel("Days")
axes[0].set_ylabel("Listening Hours")
axes[0].set_title("Spotify Listening")
axes[0].grid(True)

axes[1].plot(days, youtube_hours, marker="o")
axes[1].set_xlabel("Days")
axes[1].set_ylabel("Viewing Hours")
axes[1].set_title("YouTube Viewing")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

```python
# 4.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify_hours = [2, 3, 2.5, 4, 3, 5, 4]

youtube_hours = [1.5, 2, 2.5, 1, 3, 4, 3.5]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].plot(days, spotify_hours, marker="o")
axes[0].set_xlabel("Days")
axes[0].set_ylabel("Listening Hours")
axes[0].set_title("Weekly Spotify Listening Hours")
axes[0].grid(True)

axes[1].plot(days, youtube_hours, marker="o")
axes[1].set_xlabel("Days")
axes[1].set_ylabel("Viewing Hours")
axes[1].set_title("Weekly YouTube Viewing Hours")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```
