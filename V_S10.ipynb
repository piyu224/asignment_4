```python
# 1.

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = {
    "Day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
    "Songs": [40, 55, 50, 65, 70, 90, 80]
}

df = pd.DataFrame(data)

sns.set(context="poster")

sns.barplot(data=df, x="Day", y="Songs")

plt.xlabel("Day")
plt.ylabel("Number of Songs")
plt.title("Spotify Songs Streamed Per Day")
plt.show()
```

```python
# 2.

import seaborn as sns
import matplotlib.pyplot as plt

cuisine = [
    "Indian", "Chinese", "Italian", "Indian", "Chinese",
    "Indian", "Italian", "Indian", "Chinese", "Italian",
    "Indian", "Chinese", "Italian", "Indian", "Chinese",
    "Indian", "Italian", "Chinese", "Indian", "Italian"
]

sns.countplot(
    x=cuisine,
    hue=cuisine,
    palette="colorblind",
    legend=False
)

plt.xlabel("Cuisine Type")
plt.ylabel("Number of Orders")
plt.title("Food Orders by Cuisine Type")
plt.show()
```

```python
# 3.

import seaborn as sns
import matplotlib.pyplot as plt

days = [
    "Day 1", "Day 2", "Day 3", "Day 4", "Day 5", "Day 6", "Day 7",
    "Day 8", "Day 9", "Day 10", "Day 11", "Day 12", "Day 13", "Day 14"
]

steps = [
    4500, 5200, 6000, 5800, 7000, 7500, 6800,
    7200, 8000, 7600, 8500, 9000, 8200, 9500
]

sns.set(context="notebook", palette="muted")

sns.lineplot(x=days, y=steps, marker="o")

plt.xlabel("Day")
plt.ylabel("Number of Steps")
plt.title("Daily Step Counts - 2 Weeks")
plt.xticks(rotation=45)
plt.show()
```

```python
# 4.

import seaborn as sns
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales = [20, 35, 30, 45, 50, 60, 55]

sns.set(
    palette="colorblind",
    context="poster"
)

sns.barplot(x=days, y=sales)

plt.xlabel("Day")
plt.ylabel("Sales")
plt.title("Weekly Sales")

plt.show()
```

```python
# 5.

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Product sales data
data = {
    "Category": ["Mobiles", "Laptops", "TVs", "Headphones", "Watches"],
    "Sales": [85000, 65000, 50000, 35000, 42000]
}

df = pd.DataFrame(data)

sns.set(
    palette="deep",
    context="talk",
    style="whitegrid",
    font_scale=1.1
)

sns.barplot(data=df, x="Category", y="Sales")

plt.xlabel("Product Category")
plt.ylabel("Sales Amount")
plt.title("Flipkart Product Category Sales")
plt.xticks(rotation=20)
plt.show()
```
