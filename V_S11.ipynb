```python
# 1.

import seaborn as sns
import matplotlib.pyplot as plt

categories = ["Pizza", "Burger", "Biryani", "Momos", "Ice Cream"]
orders = [45, 60, 50, 35, 40]

sns.set_theme(style="whitegrid")

sns.barplot(x=categories, y=orders)

plt.xlabel("Food Category")
plt.ylabel("Number of Orders")
plt.title("Food Category Orders")

plt.savefig("food_orders.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# 2.

import seaborn as sns
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [4500, 5200, 6000, 5800, 7000, 7500, 8000]

sns.set_theme(style="whitegrid")

sns.lineplot(x=days, y=steps, marker="o")

plt.xlabel("Day")
plt.ylabel("Steps")
plt.title("Daily Step Counts")

plt.savefig("daily_steps.svg", format="svg", bbox_inches="tight")
plt.show()
```

```python
# 3.

import seaborn as sns
import matplotlib.pyplot as plt

products = ["Mobile", "Laptop", "TV", "Tablet", "Watch"]
sales = [80, 65, 50, 40, 55]

sns.set_theme(style="darkgrid", palette="deep")

sns.barplot(x=products, y=sales)

plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Product Sales")

plt.savefig("sales_chart.png", dpi=1200, bbox_inches="tight")
plt.show()
```

```python
# 4.

import seaborn as sns
import matplotlib.pyplot as plt

def apply_presentation_theme():
    sns.set_theme(
        style="whitegrid",
        palette="deep",
        font="Arial",
        font_scale=1.2
    )

apply_presentation_theme()

apps = ["Instagram", "YouTube", "WhatsApp", "Spotify", "Chrome"]
usage = [30, 25, 20, 15, 10]

plt.pie(
    usage,
    labels=apps,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Top 5 Most-Used Apps")

plt.savefig("top_apps.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = pd.DataFrame({
    "Math": [80, 70, 90, 85],
    "Science": [75, 85, 88, 80],
    "English": [90, 80, 85, 95],
    "Computer": [95, 90, 92, 88]
})

sns.set_theme(style="whitegrid")

sns.heatmap(data, annot=True, cmap="Blues")

plt.title("Student Marks Heatmap")

plt.savefig(
    "student_heatmap.svg",
    format="svg",
    dpi=1200,
    bbox_inches="tight"
)

plt.show()
```
