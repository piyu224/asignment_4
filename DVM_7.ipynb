```python
# 1.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

steps = [5000, 6500, 7000, 5500, 8000, 9000, 7500]
water = [2, 2.5, 2, 3, 2.5, 3, 2.5]

plt.figure(figsize=(8, 7))

plt.subplot(2, 1, 1)
plt.plot(days, steps, marker="o")
plt.title("Daily Step Count")
plt.ylabel("Steps")

plt.subplot(2, 1, 2)
plt.plot(days, water, marker="o")
plt.title("Daily Water Intake")
plt.xlabel("Days")
plt.ylabel("Water (Litres)")

plt.tight_layout()
plt.show()
```

```python
# 2.

import matplotlib.pyplot as plt

quarters = ["Q1", "Q2", "Q3", "Q4"]
posts = [25, 40, 30, 50]

fig, ax = plt.subplots(2, 2, figsize=(10, 8))

ax[0, 0].plot(quarters, posts, marker="o")
ax[0, 0].set_title("Line Chart")
ax[0, 0].set_ylabel("Posts")

ax[0, 1].bar(quarters, posts)
ax[0, 1].set_title("Bar Chart")
ax[0, 1].set_ylabel("Posts")

ax[1, 0].scatter(quarters, posts)
ax[1, 0].set_title("Scatter Chart")
ax[1, 0].set_ylabel("Posts")

ax[1, 1].pie(posts, labels=quarters, autopct="%1.1f%%")
ax[1, 1].set_title("Pie Chart")

plt.suptitle("Instagram Posts by Quarter")
plt.tight_layout()
plt.show()
```

```python
# 3.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

zomato = [8, 10, 7, 12, 9, 11]
swiggy = [6, 9, 8, 10, 12, 9]
dominos = [4, 5, 6, 7, 5, 8]

fig, ax = plt.subplots(1, 3, figsize=(12, 4), sharey=True)

ax[0].plot(months, zomato, marker="o")
ax[0].set_title("Zomato")
ax[0].set_xlabel("Months")
ax[0].set_ylabel("Number of Orders")

ax[1].plot(months, swiggy, marker="o")
ax[1].set_title("Swiggy")
ax[1].set_xlabel("Months")

ax[2].plot(months, dominos, marker="o")
ax[2].set_title("Domino's")
ax[2].set_xlabel("Months")

plt.suptitle("Food Orders Over 6 Months")
plt.tight_layout()
plt.show()
```

```python
# 4.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

youtube = [8, 10, 12, 9, 14, 15]
instagram = [5, 7, 8, 6, 9, 10]
spotify = [3, 4, 5, 4, 6, 7]
whatsapp = [2, 3, 3, 4, 3, 5]

fig, ax = plt.subplots(2, 2, figsize=(10, 8), sharex=True)

ax[0, 0].plot(months, youtube, marker="o")
ax[0, 0].set_title("YouTube")
ax[0, 0].set_ylabel("Data Usage (GB)")

ax[0, 1].plot(months, instagram, marker="o")
ax[0, 1].set_title("Instagram")
ax[0, 1].set_ylabel("Data Usage (GB)")

ax[1, 0].plot(months, spotify, marker="o")
ax[1, 0].set_title("Spotify")
ax[1, 0].set_xlabel("Months")
ax[1, 0].set_ylabel("Data Usage (GB)")

ax[1, 1].plot(months, whatsapp, marker="o")
ax[1, 1].set_title("WhatsApp")
ax[1, 1].set_xlabel("Months")
ax[1, 1].set_ylabel("Data Usage (GB)")

plt.suptitle("Monthly Mobile Data Usage")
plt.tight_layout()
plt.show()
```
