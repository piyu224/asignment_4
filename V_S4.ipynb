```python
# 1.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Documents/flipkart_products.csv")

print(df.columns)

sns.boxplot(y=df.iloc[:, 1])

plt.title("Flipkart Product Price Distribution")
plt.ylabel("Price")
plt.show()
```

```python
# 2.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Category": [
        "Fast Food", "Fast Food", "Fast Food", "Fast Food", "Fast Food",
        "Desserts", "Desserts", "Desserts", "Desserts", "Desserts"
    ],
    "Order_Amount": [250, 300, 350, 450, 500, 150, 200, 250, 300, 350]
}

df = pd.DataFrame(data)

sns.violinplot(data=df, x="Category", y="Order_Amount")

plt.title("Zomato Order Amount")
plt.xlabel("Restaurant Category")
plt.ylabel("Order Amount (₹)")

plt.show()
```

```python
# 3.

import seaborn as sns
import matplotlib.pyplot as plt

prices = [500, 700, 800, 1000, 1200, 1500, 1800, 2000, 2500, 3000, 10000]

sns.boxplot(y=prices)

plt.title("IPL Match Ticket Prices")
plt.ylabel("Ticket Price (₹)")

plt.show()
```
