```python
# 2.

import matplotlib.pyplot as plt

names = ["Amit", "Riya", "Rahul", "Priya", "Neha", "Jay",
         "Anjali", "Karan", "Pooja", "Vikas", "Sneha", "Rohan"]

orders = [5, 8, 3, 10, 6, 12, 4, 9, 7, 11, 5, 8]

amount = [800, 1200, 450, 1600, 900, 2000,
          600, 1400, 1100, 1800, 750, 1300]

plt.scatter(orders, amount)

for i in range(len(names)):
    plt.annotate(names[i], (orders[i], amount[i]))

plt.xlabel("Number of Zomato Orders")
plt.ylabel("Amount Spent (₹)")
plt.title("Zomato Orders vs Amount Spent")

plt.show()
```

```python
# 3.

import matplotlib.pyplot as plt
import numpy as np

names = ["Amit", "Riya", "Rahul", "Priya", "Neha", "Jay",
         "Anjali", "Karan", "Pooja", "Vikas", "Sneha", "Rohan"]

orders = [5, 8, 3, 10, 6, 12, 4, 9, 7, 11, 5, 8]

amount = [800, 1200, 450, 1600, 900, 2000,
          600, 1400, 1100, 1800, 750, 1300]

plt.scatter(orders, amount)

for i in range(len(names)):
    plt.annotate(names[i], (orders[i], amount[i]))

slope, intercept = np.polyfit(orders, amount, 1)

x = np.array(orders)
y = slope * x + intercept

plt.plot(x, y)

plt.xlabel("Number of Zomato Orders")
plt.ylabel("Amount Spent (₹)")
plt.title("Zomato Orders vs Amount Spent with Trend Line")

plt.show()

print("Slope =", slope)
print("Intercept =", intercept)
```

```python
# 4.

import matplotlib.pyplot as plt
import numpy as np

youtube = [5, 8, 10, 12, 7, 15, 9, 11, 6, 14, 13, 8, 16, 10, 12]

spotify = [6, 7, 9, 11, 6, 13, 8, 10, 5, 12, 11, 7, 14, 9, 10]

slope, intercept = np.polyfit(youtube, spotify, 1)

x = np.array(youtube)
y = slope * x + intercept

plt.scatter(youtube, spotify)

plt.plot(x, y)

equation = f"y = {slope:.2f}x + {intercept:.2f}"

plt.text(6, 13, equation)

plt.xlabel("YouTube Watch Hours")
plt.ylabel("Spotify Listening Hours")
plt.title("YouTube vs Spotify Usage")

plt.show()

print("Regression Equation:", equation)
```

```python
# 5.

import matplotlib.pyplot as plt
import numpy as np

orders = [2, 4, 3, 6, 5, 8, 7, 9, 4, 10]

amount = [1200, 2500, 1800, 3600, 3000,
          4500, 4000, 5200, 2300, 6000]

plt.scatter(orders, amount)

slope, intercept = np.polyfit(orders, amount, 1)

x = np.array(orders)
y = slope * x + intercept

plt.plot(x, y)

plt.xlabel("Number of Flipkart Orders")
plt.ylabel("Total Amount Spent (₹)")
plt.title("Flipkart Orders vs Total Amount Spent")

plt.show()
```
