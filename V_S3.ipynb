```python
# 1.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Delivery_Time": [
        25, 30, 35, 28, 40, 32, 27, 45, 38, 30,
        25, 35, 42, 29, 33, 37, 31, 26, 40, 36,
        34, 28, 45, 32, 39, 30, 27, 41, 35, 33
    ]
}

df = pd.DataFrame(data)

sns.histplot(data=df, x="Delivery_Time", bins=8)

plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Number of Orders")
plt.title("Distribution of Swiggy Delivery Times")
plt.show()
```

```python
# 2.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

steps = [5200, 6800, 7500, 6100, 8200,9000, 7200, 6500, 7800, 8500,5600, 7000, 9200, 8100, 6300,7400, 8800, 9500, 6900, 7600,5900, 8300, 7100, 8700, 
         6400,7900, 9100, 7300, 6800, 8600]
df = pd.DataFrame({"Steps": steps})

sns.kdeplot(data=df, x="Steps", fill=True)

plt.xlabel("Daily Steps")
plt.ylabel("Probability Density")
plt.title("Distribution of Daily Step Counts")
plt.show()
```

```python
# 3.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

spending = [250, 450, 300, 600, 750,500, 850, 400, 900, 650,550, 700, 350, 1000, 800,450, 600, 950, 500, 750,300, 850, 650, 700, 550,900, 400, 
            800, 600, 1000]
df = pd.DataFrame({"Monthly_Spending": spending})

fig, ax = plt.subplots()

sns.histplot(
    data=df,
    x="Monthly_Spending",
    bins=8,
    kde=False,
    ax=ax
)

sns.kdeplot(
    data=df,
    x="Monthly_Spending",
    ax=ax
)

plt.xlabel("Monthly Spending (₹)")
plt.ylabel("Frequency / Density")
plt.title("Zomato Monthly Spending Distribution")

plt.show()
```

```python
# 4.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

runs = [
    85, 120, 145, 160, 175, 190, 110, 135, 155, 180,
    200, 210, 95, 125, 140, 165, 185, 195, 220, 230,
    105, 130, 150, 170, 188, 205, 215, 225, 240, 250
]

df = pd.DataFrame({"Runs": runs})

sns.histplot(data=df, x="Runs", bins=5)
plt.title("IPL Run Distribution - 5 Bins")
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.show()

sns.histplot(data=df, x="Runs", bins=10)
plt.title("IPL Run Distribution - 10 Bins")
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.show()

sns.histplot(data=df, x="Runs", bins=20)
plt.title("IPL Run Distribution - 20 Bins")
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.show()
```
