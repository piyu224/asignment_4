```python
# 1.

import matplotlib.pyplot as plt

delivery_times = [32, 28, 45, 30, 36, 40, 29, 34, 38, 31,
                  27, 41, 33, 35, 39, 37, 44, 30, 29, 32,
                  36, 28, 43, 31, 35, 40, 38, 33, 30, 42]

plt.hist(delivery_times, bins=5)

plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Number of Orders")
plt.title("Swiggy Delivery Time Distribution")

plt.show()
```

```python
# 2.

import matplotlib.pyplot as plt

scores = [125, 140, 155, 180, 195, 210, 175, 160, 145, 200,
          135, 150, 170, 185, 220, 130, 165, 190, 205, 155,
          145, 175, 185, 195, 215, 125, 135, 150, 165, 180,
          200, 210, 140, 155, 170, 190, 205, 220, 130, 145,
          160, 175, 185, 200, 215, 135, 150, 165, 195, 210,
          145, 180, 205, 220, 155, 170, 190, 200, 130, 160]

plt.hist(scores, bins=5)
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.title("IPL Scores - 5 Bins")
plt.show()

plt.hist(scores, bins=10)
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.title("IPL Scores - 10 Bins")
plt.show()

plt.hist(scores, bins=15)
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.title("IPL Scores - 15 Bins")
plt.show()
```

```python
# 3.

import matplotlib.pyplot as plt
import random

steps = [random.randint(4000, 12000) for i in range(30)]

plt.hist(steps, bins=8)

plt.xlabel("Daily Steps")
plt.ylabel("Frequency")
plt.title("Daily Steps - Frequency")
plt.show()

plt.hist(steps, bins=8, density=True)

plt.xlabel("Daily Steps")
plt.ylabel("Density")
plt.title("Daily Steps - Density")
plt.show()
```

```python
# 4.

import matplotlib.pyplot as plt
import random

song_durations = [random.randint(120, 300) for i in range(50)]

plt.hist(song_durations, bins=10, color="blue")

plt.xlabel("Song Duration (seconds)")
plt.ylabel("Number of Songs")
plt.title("Spotify Song Duration Distribution")

plt.show()
```

```python
# 5.

import matplotlib.pyplot as plt

prices = [150, 250, 450, 550, 700, 900,
          1100, 1500, 1800, 2100, 2500, 3000,
          4000, 4800, 350, 650, 1200, 2000, 5000]

bins = [100, 500, 1000, 2000, 5000]

plt.hist(prices, bins=bins, edgecolor="black")

plt.xlabel("Product Price")
plt.ylabel("Number of Products")
plt.title("Flipkart Product Price Distribution")

plt.show()
```
